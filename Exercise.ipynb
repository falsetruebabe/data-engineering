## 1. 실습 문제: 시험 점수 처리 (Operation with Test Scores)

```python
import numpy as np

# [1단계] 배열 선언
test_1 = np.array([92, 94, 88, 91, 87])
test_2 = np.array([79, 100, 86, 93, 91])
test_3 = np.array([87, 85, 72, 90, 92])

print(test_1)
print(test_2)
print(test_3)
```

```python
# [2단계] 배열 연산 (Broadcasting) - 세 번째 시험 점수에 각각 +2점

test_3_fixed = test_3 + 2

test_3_fixed
```

```python
test_3 = test_3 + 2

test_3
```

```python
total_grade = test_1 + test_2 + test_3_fixed

total_grade
```

```python
final_grade = total_grade / 3

final_grade
```
