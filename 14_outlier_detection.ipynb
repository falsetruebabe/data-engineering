## **14. 이상치 탐지**

```python
import pandas as pd
import numpy as np
from datetime import datetime
from matplotlib import pyplot as plt
```

### **데이터 탐색**

```python
"""
    데이터셋 로딩
"""

def parser(x):
    return datetime.strptime(x, '%Y-%m-%d %H:%M:%S')

path = './data/'
file = 'AirQualityUCI_refined.csv'

df = pd.read_csv(
    path + file,
    index_col=[0],
    parse_dates=[0],
    date_parser=parser
)

df.head()
```

```python
# [+] 일산화탄소 변수 (CO(GT)) 시각화
df.info()
```

```python
# [+] 결측 데이터 처리: 선형 보간
co = df['CO(GT)'].copy() # Series 복사
co.interpolate(method='linear', inplace=True) # 선형 보간
```

```python
# 결측치 처리 결과 시각화
plt.plot(df['CO(GT)'], label='original', zorder=2)
plt.plot(co, label='linear interpolation', zorder=1)
plt.legend(loc='best')
plt.show()
```

```python
# [+] 이상치 시각화: 박스플롯
plt.boxplot(co)
plt.xlabel('CO(GT)')
plt.ylabel('Concentration (${mg/m^3}$)')
plt.show()
```

```python
# [+] 변수 간 상관관계 측정
corr_matrix = df.corr()
corr_matrix
```

```python
# 일산화탄소와 가장 상관관계가 약한 변수 선택
rh = df['RH'].copy().interpolate() # RH: 상대습도
```

```python
# 산포도 시각화
plt.scatter(co, rh, s=12, c='black')
plt.xlabel('CO(GT)')
plt.ylabel('RH')
plt.show()
## 상관계수가 낮다.
```

```python
# 일산화탄소와 가장 상관관계가 강한 변수 선택
nmhc = df['PT08.S2(NMHC)'].copy().interpolate() # NMHC: 비메탄계 탄화수소
```

```python
# 산포도 시각화
plt.scatter(co, nmhc, s=12, c='black')
plt.xlabel('CO(GT)')
plt.ylabel("NMHC")
plt.show()
```

### **사분위수 범위 기반 이상치 탐지**

```python
# [+] Q1, Q2, Q3 계산
q1 = co.quantile(0.25) ## 하위 25% = Q1
q2 = co.quantile(0.5) ## 하위 50% = Q2
q3 = co.quantile(0.75) ## 하위 75% = Q3
print(q1, q2, q3)
```

```python
co
```

```python
# [+] IQR, 상한(upper_fence), 하한(lower_fence)
iqr = q3 - q1
upper_fence = q3 + 1.5*iqr
lower_fence = q1 - 1.5*iqr
print(iqr, upper_fence, lower_fence)
```

```python
# [+] 이상치 선택
outliers = co.loc[(co > upper_fence) | (co < lower_fence)]
outliers
```

```python
co.index.isin(outliers.index) ## 이상치 여부를 마스킹하는 코드
```

```python
# [+] 이상치 여부 마스킹
mask = co.index.isin(outliers.index) 
mask[:50]
```

```python
len(co)
```

```python
mask.sum() / len(co)
```

```python
# 정상 데이터 / 이상치 시각화
plt.plot(co[~mask], label='normal', color='blue',
    marker='o', markersize=3, linestyle='None')
plt.plot(outliers, label='outliers', color='red',
    marker='x', markersize=3, linestyle='None')
plt.legend(loc='best')
```

```python
# [+] 이상치 제거
co_refined = co.copy()
co_refined[mask] = np.nan ## 결측치로 바꿔서 이상치 제거
```

```python
co_refined.plot()
```

```python
# [+] 제거된 이상치에 대한 대치: 선형보간
co_refined.interpolate(inplace=True)
co_refined.plot()
```

### **Z-점수 기반 이상치 탐지**

```python
# 일산화탄소 변수 분포 시각화
import seaborn as sns
sns.distplot(co)
```

```python
# [+] 평균, 표준편차
mean = co.mean()
std = co.std()
## mean = np.mean(co)
## std = np.std(co)
print(mean, std)
```

```python
# [+] 데이터 포인트별 Z-점수 계산
outliers = []
thres = 3 ## 3이 가장 일반적임

for i in co:
    z_score = (i - mean) / std
    if np.abs(z_score) > thres: ## +- 3이기때문에 절댓값 씌움
        print(z_score)
        outliers.append(i)
```

```python
# [+] 이상치 선택
outliers = co.loc[(np.abs(co-mean) / std) > 3]
outliers
```

```python
# [+] 마스킹
mask = co.index.isin(outliers.index) ## True: 이상치  False: 정상
mask[:50]
```

```python
# 이상치 제거 전/후 분포 시각화
sns.distplot(co, axlabel='CO(GT)', label='original')
sns.distplot(co[~mask], label='outliers removed') 
## ~ = not 연산자 i.e. 정상데이터만 선택(이상치제거)
plt.legend(loc='best')
plt.show()
```

```python
# 기타 이상치 제거 방법: Flooring, Capping
floor = co.quantile(0.1) ## 작은 값을 가진 이상치를 적당히 작은 값으로 대체함
cap = co.quantile(0.9) ## 반대로 큰값을 가지는 값을 적당히 큰 값으로 대체함

co.loc[co < floor] = floor
co.loc[co > cap] = cap
```

```python
# 시각화
co.plot()
```
